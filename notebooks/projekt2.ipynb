```python
def find_next_connection(delayed_trains_df, departure_trains_df, min_transfer_minutes=5):
    """
    Findet für jeden verspäteten Zug den nächstmöglichen Anschlusszug (planmäßige Abfahrt).
    """
    connection_results = []
    min_transfer_time = pd.Timedelta(minutes=min_transfer_minutes)

    # Sortiere alle potenziellen Anschlüsse nach planmäßiger Abfahrtszeit
    departure_trains_df = departure_trains_df.sort_values(by='departure_plan').reset_index(drop=True)

    # Iteriere durch jeden verspäteten Ankunftszug
    for index, arr_row in delayed_trains_df.iterrows():
        arrival_changed = arr_row['arrival_change']
        earliest_departure_plan = arrival_changed + min_transfer_time

        # Suche den ersten Zug, dessen planmäßige Abfahrt >= der frühestmöglichen Abfahrtszeit ist
        next_connection_candidates = departure_trains_df[
            departure_trains_df['departure_plan'] >= earliest_departure_plan
        ]
        
        # Prüfung, ob ein Anschluss gefunden wurde
        if not next_connection_candidates.empty:
            conn_row = next_connection_candidates.iloc[0]
            
            # Wartezeit berechnen (Zeit zwischen tatsächlicher Ankunft und planmäßiger Abfahrt des Anschlusses)
            wait_time = conn_row['departure_plan'] - arrival_changed
            
            # Zielbahnhof extrahieren (letzter Halt in der path_list oder die Zielstation selbst)
            final_destination = conn_row['path_list'][-1] if isinstance(conn_row['path_list'], list) and conn_row['path_list'] else conn_row['station']

            connection_results.append({
                'Delay_Ankunft_Geändert': arrival_changed,
                'Delay_Min': arr_row['arrival_delay_calc'],
                'Delay_Kategorie': arr_row['category'],
                
                'Anschluss_Kategorie': conn_row['category'],
                'Anschluss_Zielbahnhof': final_destination,
                'Verpasste_Verbindung': (conn_row['departure_plan'] - arr_row['arrival_plan']).total_seconds() / 60 < min_transfer_minutes,
                'Naechste_Abfahrt': conn_row['departure_plan'],
                'Wartezeit_Min': wait_time.total_seconds() / 60
            })

    return pd.DataFrame(connection_results)

# Ausführung der Analyse
connection_analysis_df = find_next_connection(df_leipzig_synthetic, df_leipzig_departures)
```

```python
def analyze_missed_and_next_connections(delayed_trains_df, departure_trains_df, min_transfer_minutes=5):
    """
    Findet für jeden verspäteten Zug den verpassten Anschluss (falls vorhanden) und den nächsten Anschluss.
    """
    results = []
    min_transfer_time = pd.Timedelta(minutes=min_transfer_minutes)
    
    # Sortiere Abfahrten einmalig nach Planzeit
    departure_trains_df = departure_trains_df.sort_values(by='departure_plan').reset_index(drop=True)

    for index, arr_row in delayed_trains_df.iterrows():
        arr_plan = arr_row['arrival_plan']
        arr_changed = arr_row['arrival_change']
        
        # 1. Bestimme den letzten potenziell erreichbaren Zug (Z_verpasst)
        # Planmäßige Abfahrt liegt nach planmäßiger Ankunft + Mindestumstiegszeit
        # UND planmäßige Abfahrt liegt vor tatsächlicher Ankunft + Mindestumstiegszeit
        
        # Züge, die planmäßig erreichbar waren
        potentially_reachable = departure_trains_df[
            departure_trains_df['departure_plan'] >= arr_plan + min_transfer_time
        ]

        # Züge, die tatsächlich verpasst wurden (Abfahrt vor tatsächlicher Ankunft + Puffer)
        missed_candidates = potentially_reachable[
            potentially_reachable['departure_plan'] < arr_changed + min_transfer_time
        ]
        
        # 2. Bestimme den nächstmöglichen Zug (Z_next)
        earliest_departure_plan = arr_changed + min_transfer_time
        next_candidates = departure_trains_df[
            departure_trains_df['departure_plan'] >= earliest_departure_plan
        ]

        missed_conn = missed_candidates.iloc[-1] if not missed_candidates.empty else None
        next_conn = next_candidates.iloc[0] if not next_candidates.empty else None
        
        # Extrahiere Daten für den verpassten Zug
        missed_data = {
            'Verpasst_ID': missed_conn['line'] if missed_conn is not None else np.nan,
            'Verpasst_Kat': missed_conn['category'] if missed_conn is not None else 'Kein direkter Anschluss verpasst',
            'Verpasst_Ziel': missed_conn['path_list'][-1] if missed_conn is not None and isinstance(missed_conn['path_list'], list) else (missed_conn['station'] if missed_conn is not None else np.nan),
            'Verpasst_Abfahrt_Plan': missed_conn['departure_plan'] if missed_conn is not None else pd.NaT,
        }
        
        # Extrahiere Daten für den nächsten Zug und berechne Wartezeit
        next_data = {
            'Naechste_Abfahrt': next_conn['departure_plan'] if next_conn is not None else pd.NaT,
            'Naechste_Wartezeit_Min': ((next_conn['departure_plan'] - arr_changed).total_seconds() / 60) if next_conn is not None else np.nan,
        }

        results.append({
            'Verspätung_Min': arr_row['arrival_delay_calc'],
            'Verspätung_Kat': arr_row['category'],
            **missed_data,
            **next_data
        })

    return pd.DataFrame(results)

# Führen Sie die erweiterte Analyse aus
missed_connections_df = analyze_missed_and_next_connections(df_leipzig_synthetic, df_leipzig_departures)
```
